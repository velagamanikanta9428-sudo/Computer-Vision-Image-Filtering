```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def show_comparison(original,blur3,blur5,blur7):
  plt.figure(figsize=(12,10))
  plt.subplot(2,2,1)
  plt.imshow(original,cmap="gray")
  plt.title('Original Grayscale Image')
  plt.axis('off')
  #3x3
  plt.subplot(2,2,2)
  plt.imshow(blur3,cmap='gray')
  plt.title('average filter(3x3)')
  plt.axis('off')
  #5x5
  plt.subplot(2,2,3)
  plt.imshow(blur5,cmap='gray')
  plt.title('average filter(5x5)')
  plt.axis('off')
  #7x7
  plt.subplot(2,2,4)
  plt.imshow(blur7,cmap='gray')
  plt.title('average filter(7x7)')
  plt.axis('off')
  plt.tight_layout()
  plt.show()
```

```python
img=cv2.imread('/content/drive/MyDrive/COMPUTER VISION/image16.jpg',0)
```

```python
if img is None:
  print('Error loading image')
else:
  #define averaging
  blur3=np.ones((3,3),np.float32)/9
  blur5=np.ones((5,5),np.float32)/25
  blur7=np.ones((7,7),np.float32)/49
```

```python
blur3x3=cv2.filter2D(img,-1,blur3)
blur5x5=cv2.filter2D(img,-1,blur5)
blur7x7=cv2.filter2D(img,-1,blur7)
```

```python
show_comparison(img,blur3x3,blur5x5,blur7x7)
```
